```python
import re

def convert_to_seconds(value):
    value = str(value).strip()

    minutes = re.search(r"(\d+)\s*min", value)
    seconds = re.search(r"(\d+)\s*s", value)

    if minutes:
        total_seconds = int(minutes.group(1)) * 60

        if seconds:
            total_seconds += int(seconds.group(1))

        return total_seconds

    return None
```

```python
# 1. Clean Congestion
for df in [congestion_city_long, congestion_urban_long]:
    df["Congestion"] = (
        df["Congestion"]
        .str.replace("%", "", regex=False)
        .astype(float)
    )

# 2. Clean Speed
for df in [speed_city_long, speed_urban_long]:
    df["Speed_kmh"] = (
        df["Speed_kmh"]
        .str.replace(" km/h", "", regex=False)
        .astype(float)
    )

# 3. Rebuild City data
city_data = congestion_city_long.merge(
    speed_city_long,
    on=["Time", "Time_24", "Day"],
    how="inner"
).merge(
    time_city_long,
    on=["Time", "Time_24", "Day"],
    how="inner"
)

city_data["Area"] = "City"

# 4. Rebuild Urban data
urban_data = congestion_urban_long.merge(
    speed_urban_long,
    on=["Time", "Time_24", "Day"],
    how="inner"
).merge(
    time_urban_long,
    on=["Time", "Time_24", "Day"],
    how="inner"
)

urban_data["Area"] = "Urban"

# 5. Combine City + Urban
master_data = pd.concat(
    [city_data, urban_data],
    ignore_index=True
)

# 6. Convert Travel Time to seconds
master_data["Travel_Time_Seconds"] = (
    master_data["Travel_Time"].apply(convert_to_seconds)
)
```
